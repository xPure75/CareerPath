```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('synthetic_job_postings.csv')
df = df.drop(df[df['posting_id'] == 212].index)
df = df.drop(df[df['posting_id'] == 703].index)

df.info()
```

```python
df['salary_usd'].describe()
```

```python
plt.figure(figsize=(8,5))
plt.hist(df['salary_usd'].dropna(), bins=20, edgecolor='black')
plt.title('Distribution of Salaries')
plt.xlabel('Salary (USD)')
plt.ylabel('Number of Job Listings')
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(x='experience_level',
            y='salary_usd',
            data=df)

plt.title('Salary by Experience Level')
plt.xlabel('Experience Level')
plt.ylabel('Salary (USD)')
plt.show()
```
